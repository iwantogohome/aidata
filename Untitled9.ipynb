```python
import re

tweet = 'Good Morning! RT @PythonUser I like Python #Python'
pattern = re.compile(r'[@#]\w+|RT|CC')
message = re.sub(pattern, '', tweet)
print(message)
```

```python
def format_date(date):
    m, d, y = date.split('-')
    return '/'.join((y, m, d))

user_date = input('mm-dd-yyyy 형식의 날짜를 입력하시오: ')
print(f'출력 결과: {format_date(user_date)}')
```

```python
text='To love and win is the best thing. To love and lose, is the next best thing.'
wordDic={}

for word in text.lower().split():
    if word in wordDic:
        wordDic[word] += 1
    else:
        wordDic[word] = 1

for word, count in sorted(wordDic.items()):
    print(f'{word:<8} 반복 횟수: {count}회')
```
